```python
import cv2
import numpy
```

```python
# Load an image in grayscale mode
im_g = cv2.imread('smallgray.png', 0) # 0 = grayscale, 1 = color image

# Display the grayscale image
im_g
```

```python
# Load an image in color mode
im_g2 = cv2.imread('smallgray.png', 1) # 0 = grayscale, 1 = color image
print(im_g2)
```

```python
# Creating img out of numpy array
cv2.imwrite("smallgray.png", im_g)
```

```python
# Display first two rows of pixels
im_g[0:2]
```

```python
# Display first two rows and first two columns of pixels
im_g[0:2, 2:4]
```

```python
# Display rows 2 to 4 and columns 2 to 4 of pixels
im_g[:, 2:4]
```

```python
# Display pixel value at row 2, column 4
im_g[2,4]
```

```python
# Iterate through each pixel in the grayscale image and print its value
for i in im_g:
    print(i)
```

```python
# Iterate through each pixel in the grayscale image and print its value
for i in im_g.T:    #T= transpose
    print(i)
```

```python
# Iterate through each pixel in the grayscale image and print its value
for i in im_g.flat:
    print(i)
```

```python
# Create a new image by stacking three copies of the grayscale image horizontally
ims = numpy.hstack((im_g, im_g, im_g))
print(ims)
```
