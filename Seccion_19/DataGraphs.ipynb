```python
# Linea de grafico basico
# Importar las bibliotecas necesarias
from bokeh.plotting import figure
from bokeh.io import output_file, show
import pandas

#Datos
df = pandas.read_csv("../Seccion_19/data/data.csv")
x = df["x"]
y = df["y"]

# Crear el archivo de salida
output_file("../Seccion_19/Graficos/BasicDataGraph.html")

# Crear el objeto figura
f = figure()

# Agregar la linea al grafico
f.line(x, y)

# Mostrar el grafico
show(f)
```

```python
# Gráfico de porcentaje de mujeres con título de licenciatura en Ingeniería por año
from bokeh.plotting import figure
from bokeh.io import output_file, show
import pandas

# Leer los datos del archivo CSV de licenciaturas
df = pandas.read_csv("../Seccion_19/data/bachelors.csv")
x = df["Year"]
y = df["Engineering"]

# Crear el archivo de salida
output_file("../Seccion_19/Graficos/EngineeringBachelors.html")

# Crear el objeto figura con título y etiquetas de ejes
f = figure(title="Porcentaje de Mujeres con Licenciatura en Ingeniería (EE.UU.)", 
           x_axis_label="Año", 
           y_axis_label="Porcentaje (%)",
           width=700, 
           height=700)

# Agregar la línea al gráfico
f.line(x, y, line_width=2, color='blue')

# Mostrar el gráfico
show(f)
```

```python
# Gráfico de dispersión - Temperatura vs Presión
from bokeh.plotting import figure
from bokeh.io import output_file, show
import pandas

# Leer datos del archivo Excel
df = pandas.read_excel("../Seccion_19/data/verlegenhuken.xlsx")

# Extraer las columnas de temperatura y presión
df["Temperature"]=df["Temperature"]/10
df["Pressure"]=df["Pressure"]/10

# Crear el archivo de salida
output_file("../Seccion_19/Graficos/scatterPlot.html")

# Crear el objeto figura con título y etiquetas de ejes
f = figure(title="Relación entre Temperatura y Presión", 
           x_axis_label="Temperature (°C)", 
           y_axis_label="Pressure (hPa)",
           width=500, 
           height=400)

# Personalizar el estilo del gráfico
f.title.text_font_size="20pt"
f.title.text_color="Gray"
f.title.text_font="arial"
f.title.text_font_style="bold"
f.xaxis.minor_tick_line_color=None
f.yaxis.minor_tick_line_color=None

# Crear scatter plot (puntos de dispersión)
f.circle(df["Temperature"],df["Pressure"],size=0.5)

# Mostrar el gráfico
show(f)
```

```python
#https://www.google.com/finance/quote/ADBE:NASDAQ?startdate=Jan+01,+2009&enddate=Aug+2,+2012&output=csv

from bokeh.plotting import figure
from bokeh.io import output_file, show
import pandas
import yfinance as yf

# Descargar datos de Adobe (ADBE)
df = yf.download('ADBE', start='2009-01-01', end='2012-08-02')

df = df.reset_index()

# Opción 1: Usar sizing_mode
p = figure(title="Adobe Stock Prices (2009-2012)",
           x_axis_label="Date", 
           y_axis_label="Price (USD)",
           x_axis_type="datetime",
           sizing_mode="stretch_width",  # En lugar de responsive=True
           height=400)

# Opción 2: Usar sizing_mode con diferentes valores
# sizing_mode="fixed"          # Tamaño fijo
# sizing_mode="stretch_width"  # Se adapta al ancho
# sizing_mode="stretch_height" # Se adapta a la altura  
# sizing_mode="stretch_both"   # Se adapta a ambos
# sizing_mode="scale_width"    # Escala manteniendo proporción

p.line(df["Date"], df["Close"],color='green', alpha=0.5)

output_file("../Seccion_19/Graficos/Stock_Price.html")

show(p)
```
