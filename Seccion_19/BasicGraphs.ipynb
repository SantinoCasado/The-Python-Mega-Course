```python
# Linea de grafico basico
# Importar las bibliotecas necesarias
from bokeh.plotting import figure
from bokeh.io import output_file, show

#Datos
x = [1, 2, 3, 4, 5]
y = [6, 7, 8, 9, 10]

# Crear el archivo de salida
output_file("../Seccion_19/Graficos/line.html")

# Crear el objeto figura
f = figure()

# Agregar la linea al grafico
f.line(x, y)

# Mostrar el grafico
show(f)
```

```python
#Snippet producing the circle based plot

#Making a basic Bokeh line graph
 
#importing Bokeh
from bokeh.plotting import figure
from bokeh.io import output_file, show
 
#prepare some data
x=[3,7.5,10]
y=[3,6,9]
 
#prepare the output file
output_file("../Seccion_19/Graficos/Circles.html")
 
#create a figure object
f1=figure()
 
#create line plot
f1.circle(x,y, radius=0.2, color="blue", fill_color="blue")
 
#write the plot in the figure object
show(f1)
```

```python
import pandas
from bokeh.plotting import figure, output_file, show
 
p=figure(width=500,height=400, tools='pan')

# Tipos de personalización
p.title.text="Cool Data"                    # Título del gráfico
p.title.text_color="Gray"                   # Color del título
p.title.text_font_size="20pt"               # Tamaño de la fuente del título
p.title.text_font="times"                   # Fuente del título
p.title.text_font_style="bold"              # Estilo de la fuente del título
p.xaxis.minor_tick_line_color=None          # Quitar las marcas menores del eje x
p.yaxis.minor_tick_line_color=None          # Quitar las marcas menores del eje y
p.xaxis.axis_label="Date"                   # Etiqueta del eje x
p.yaxis.axis_label="Intensity"              # Etiqueta del eje y
p.xaxis.axis_label_text_color="Gray"        # Color de la etiqueta del eje x
p.yaxis.axis_label_text_color="Gray"        # Color de la etiqueta del eje y
p.xaxis.axis_label_text_font_size="14pt"    # Tamaño de la fuente de la etiqueta del eje x
p.yaxis.axis_label_text_font_size="14pt"    # Tamaño de la fuente de la etiqueta del eje y


p.line([1,2,3],[4,5,6])
output_file("../Seccion_19/Graficos/ColorGraph.html")
show(p)
```

```python
from bokeh.plotting import figure, output_file, show

p = figure(width=700, height=600, tools = 'pan, reset')

p.title.text = "Earthquakes"
p.title.text_color = "Orange"
p.title.text_font = "times"
p.title.text_font_style = "italic"
p.yaxis.minor_tick_line_color = "Yellow"
p.xaxis.axis_label = "Times"
p.yaxis.axis_label = "Value"

p.circle([1,2,3,4,5], [5,6,5,5,3], size = [i*2 for i in [8,12,14,15,20]], color="red", alpha=0.5)

output_file("../Seccion_19/Graficos/Scatter_plotting.html")

show(p)
```

```python
# Multiples graphos en una sola figura
from bokeh.plotting import figure, output_file, show

p = figure(width=700, height=600, tools = 'pan, reset')

p.title.text = "Earthquakes"
p.title.text_color = "Orange"
p.title.text_font = "times"
p.title.text_font_style = "italic"
p.yaxis.minor_tick_line_color = "Yellow"
p.xaxis.axis_label = "Times"
p.yaxis.axis_label = "Value"

p.circle([1,2,3,4,5], [5,6,5,5,3], size = [i*2 for i in [8,12,14,15,20]], color="red", alpha=0.5)
p.line([1,2,3,4,5], [6,7,5,6,4], line_width=2, color='blue')

output_file("../Seccion_19/Graficos/Multiple_Graphs.html")

show(p)
```
