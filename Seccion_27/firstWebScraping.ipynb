```python
import requests
from bs4 import BeautifulSoup
```

### GUARDO ESTE ARCHIVO PARA PRUEBAS DE WEB SCRAPING

```python
r = requests.get('https://pythonizing.github.io/data/example.html')
c = r.content
```

### PARSEO EL CONTENIDO DE LA PAGINA

```python
soup = BeautifulSoup(c, 'html.parser')
print(soup.prettify())
```

### EXTRAIGO LAS DIVS PERO LLAMO SOLO LA PRIMERA

```python
all = soup.find_all('div', {'class': 'cities'})
all[0]
```

### EXTRAIGO EL NOMBRE DE LA PRIMERA CIUDAD

```python
firstCity = all[0].find_all('h2')[0].text
print(firstCity)
```

### EXTRAIGO TODOS LOS NOMBRES Y DESCRIPCIONES DE LAS CIUDADES

```python
for item in all:
    cityName = item.find_all('h2')[0].text
    description = item.find_all('p')[0].text
    print(cityName + ' - ' + description)
```
